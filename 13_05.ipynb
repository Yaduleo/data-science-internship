```python
# Creating two subsets: first 5 rows and last 5 rows
top_5 = df.head(5)
bottom_5 = df.tail(5)

# Stacking them vertically (standard use for concat)
combined_vertical = pd.concat([top_5, bottom_5], axis=0)
combined_vertical
```

```python
# Create a small lookup table for regions
region_notes = pd.DataFrame({
    'region': ['North America', 'Africa'],
    'region_priority': ['High', 'Emerging']
})

# Merge this with your original data based on the 'region' column
merged_df = df.merge(region_notes, on='region', how='left')
merged_df[['country', 'region', 'region_priority']].head()
```

```python
# Create a separate dataframe with the same index
extra_info = pd.DataFrame({
    'is_verified': [True] * len(df)
}, index=df.index)

# Join based on index
joined_df = df.join(extra_info)
joined_df[['record_id', 'country', 'is_verified']].head()
```

```python
'''conditional search'''
a=df[df['govt_ai_policy_score_1_to_10']>6]
a.head()
```

```python
# Convert the 'country' column to all uppercase
df['country_upper'] = df['country'].str.upper()

# View the result
df[['country', 'country_upper']].head()
```

```python
# Convert the 'region' column to all lowercase
df['region_lower'] = df['region'].str.lower()

# View the result
df[['region', 'region_lower']].head()
```
