```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
```

```python
'''
student mark analysis
give values for x and y as random integers
show box plot
6 students, 6 marks, x axis, y axis, legend
also include columns of roll no, name of student,  and whether they passed or failed
use seaborn, not matplotlib
'''
```

```python
num_students = 6
roll_no = np.arange(1, num_students + 1)

student_names = ['Yadu','Rithu','Elakkiya','Muth', 'Maithi','Madhan']

# Generate random marks between 0 and 100
marks = np.random.randint(0, 101, num_students)

# Define passing threshold
passing_threshold = 50
passed_failed = ['Pass' if mark >= passing_threshold else 'Fail' for mark in marks]

# Create a DataFrame with student names
df_students = pd.DataFrame({
    'Roll No': roll_no,
    'Student Name': student_names,
    'Marks': marks,
    'Status': passed_failed
})

display(df_students)
```

```python
sns.boxplot(df_students['Marks'])
```

```python
# Create a box plot using seaborn
plt.figure(figsize=(8, 6))
sns.boxplot(x=df_students['Status'], y=df_students['Marks'], hue=df_students['Status'], palette={'Pass': 'green', 'Fail': 'red'}, legend='full')

plt.title('Student Mark Analysis: Marks Distribution by Pass/Fail Status')
plt.xlabel('Status')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Create a countplot to show the pass/fail distribution
plt.figure(figsize=(6, 4))
sns.countplot(x='Status', data=df_students, palette={'Pass': 'green', 'Fail': 'red'}, hue='Status', legend=False)

# Add labels and title
plt.title('Number of Students Who Passed vs. Failed')
plt.xlabel('Status')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Display the plot
plt.show()
```

```python
# Set the figure size
plt.figure(figsize=(8, 5))

# Create a bar plot to compare marks across students
sns.barplot(
    x='Student Name',
    y='Marks',
    data=df_students,
    hue='Status',
    palette={'Pass': 'green', 'Fail': 'red'}
)

# Add a horizontal line indicating the passing threshold (50 marks)
plt.axhline(y=50, color='blue', linestyle='--', alpha=0.8, label='Passing Threshold (50)')

# Add labels, title, and formatting
plt.title('Comparison of Students\' Marks')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.ylim(0, 105)  # Set limit slightly above 100 for better spacing
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.legend(title='Status')

# Display the comparison plot
plt.show()
```

```python
# 1. Violin Plot: Combines box plot features with kernel density estimation to show the distribution of marks.
plt.figure(figsize=(8, 6))
sns.violinplot(x='Status', y='Marks', data=df_students, hue='Status', palette={'Pass': 'green', 'Fail': 'red'}, inner='quartile', legend='full')
plt.title('Violin Plot of Marks by Status')
plt.xlabel('Status')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python

# 2. Swarm Plot: Shows individual data points, avoiding overlap, making it easy to see the exact mark of each student.
plt.figure(figsize=(8, 6))
sns.swarmplot(x='Status', y='Marks', data=df_students, hue='Status', palette={'Pass': 'green', 'Fail': 'red'}, size=8, legend='full')
plt.title('Swarm Plot of Marks by Status (Individual Student Marks)')
plt.xlabel('Status')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(6, 5))
# Jitter spreads out the points slightly so they don't overlap directly
sns.stripplot(x='Status', y='Marks', data=df_students, hue='Status',
              palette={'Pass': 'green', 'Fail': 'red'}, size=10, jitter=0.1)

# Annotate student names next to their points
for i in range(len(df_students)):
    status_idx = 0 if df_students['Status'][i] == 'Pass' else 1
    plt.text(status_idx + 0.05, df_students['Marks'][i], df_students['Student Name'][i], fontsize=9)

plt.title('Individual Student Marks Distribution')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.pointplot(x='Status', y='Marks', data=df_students, hue='Status',
              palette={'Pass': 'green', 'Fail': 'red'}, markers="o", linestyles="")

plt.title('Comparison of Pass vs. Fail Groups')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Sort DataFrame by marks for a cleaner ranking display
df_sorted = df_students.sort_values(by='Marks', ascending=False)

plt.figure(figsize=(7, 5))
sns.barplot(x='Marks', y='Student Name', data=df_sorted, hue='Status',
            palette={'Pass': 'green', 'Fail': 'red'}, dodge=False)

# Add passing threshold vertical line
plt.axvline(x=50, color='blue', linestyle='--', alpha=0.8, label='Passing Threshold (50)')

plt.title('Ranked Student Performance')
plt.xlabel('Marks')
plt.ylabel('Student Name')
plt.legend(title='Status')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```

```python
# 1. Grouped Bar Chart: Compare average marks of 'Pass' vs. 'Fail' students.
plt.figure(figsize=(7, 5))
sns.barplot(x='Status', y='Marks', data=df_students, hue='Status', palette={'Pass': 'green', 'Fail': 'red'}, errorbar=None, legend='full')
plt.title('Average Marks by Status')
plt.xlabel('Status')
plt.ylabel('Average Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

```

```python
# 2. Scatter Plot with Passing Threshold: Shows individual marks and the passing line.
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Student Name', y='Marks', data=df_students, hue='Status', style='Status',
                palette={'Pass': 'green', 'Fail': 'red'}, s=100, legend='full') # s is marker size
plt.axhline(y=50, color='blue', linestyle='--', alpha=0.7, label='Passing Threshold (50)')
plt.title('Individual Student Marks with Passing Threshold')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.ylim(0, 105)
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Status')
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
sns.catplot(x='Status', y='Marks', data=df_students, hue='Status', palette={'Pass': 'green', 'Fail': 'red'})
plt.title('Catplot of Marks by Status')
plt.xlabel('Status')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.kdeplot(x='Marks', hue='Status', data=df_students, fill=True, common_norm=False, palette={'Pass': 'green', 'Fail': 'red'})
plt.axvline(x=50, color='blue', linestyle='--', alpha=0.7, label='Passing Threshold (50)')
plt.title('KDE Plot of Marks by Status (Pass/Fail)')
plt.xlabel('Marks')
plt.ylabel('Density')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
df=pd.read_csv('Teen_Mental_Health_Dataset.csv')
df.head()
```

```python
plt.figure(figsize=(7, 4))
# Create the countplot for platform usage
sns.countplot(x='platform_usage', data=df, palette='Set2', hue='platform_usage', legend=False)

# Formatting
plt.title('Distribution of Social Media Platform Usage')
plt.xlabel('Platform Used')
plt.ylabel('Number of Teens')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(8, 5))
# Breakdown social interaction levels across genders
sns.countplot(x='gender', hue='social_interaction_level', data=df, palette='viridis')

# Formatting
plt.title('Social Interaction Levels by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Social Interaction')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='depression_label', data=df, palette='coolwarm', hue='depression_label', legend=False)

plt.title('Distribution of Depression Labels')
plt.xlabel('Depression Label (0 = No, 1 = Yes)')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Calculate percentages instead of raw counts
sns.histplot(
    data=df,
    x='platform_usage',
    hue='depression_label',
    multiple='fill',  # This stretches the bars to 100%
    stat='proportion',
    palette='coolwarm',
    shrink=0.8
)

plt.title('Proportion of Depression Labels across Social Media Platforms')
plt.xlabel('Platform Used')
plt.ylabel('Proportion (0.0 = 0%, 1.0 = 100%)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Creates separate count plots side-by-side for Low, Medium, and High interaction
g = sns.catplot(
    data=df,
    x='platform_usage',
    hue='depression_label',
    col='social_interaction_level', # Splits into columns
    kind='count',
    palette='Set1',
    height=4,
    aspect=0.8
)

g.set_titles("{col_name} Social Interaction")
g.set_axis_labels("Platform", "Count")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Order the stress levels from highest count to lowest count
stress_order = df['stress_level'].value_counts().index

sns.countplot(
    y='stress_level',  # Assigning to 'y' automatically makes it horizontal
    data=df,
    order=stress_order,
    palette='plasma',
    hue='stress_level',
    legend=False
)

plt.title('Ranked Distribution of Teen Stress Levels')
plt.ylabel('Stress Level (Scale)')
plt.xlabel('Number of Teens')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```
