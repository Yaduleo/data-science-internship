```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style of the plot
sns.set_theme(style="whitegrid")

# Create a line chart tracking the average AI Adoption Index over the years
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='year', y='ai_adoption_index', errorbar=None, marker='o', linewidth=2.5)

# Adding titles and labels
plt.title('Global Average AI Adoption Index Trend (2020–2026)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average AI Adoption Index', fontsize=12)

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")

# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='ai_adoption_index',
    y='sector_automation_risk_score',
    alpha=0.7
)

# Customizing titles and labels
plt.title('AI Adoption Index vs. Sector Automation Risk Score', fontsize=14, fontweight='bold')
plt.xlabel('AI Adoption Index', fontsize=12)
plt.ylabel('Sector Automation Risk Score', fontsize=12)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Scatter plot comparing female workforce share vs female displacement share
sns.scatterplot(
    data=df,
    x='pct_workforce_female',
    y='pct_displaced_roles_female',
    hue='industry_sector',
    alpha=0.4,  # High transparency helps handle the density
    palette='tab10'
)

# Add a 45-degree reference line (points above this line mean women are disproportionately displaced)
plt.plot([0, 1], [0, 1], color='red', linestyle='--', linewidth=2, label='Equal Proportion Line')

plt.title('Female Workforce Share vs. Share of Female Displaced Roles', fontsize=14, fontweight='bold')
plt.xlabel('% Workforce Female', fontsize=12)
plt.ylabel('% Displaced Roles Female', fontsize=12)
plt.legend(title='Industry Sector', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# 1. Calculate sorted order by net workforce change
income_order = df.groupby('income_group')['net_workforce_change_pct'].mean().sort_values(ascending=True).index

# 2. Create a vertical bar chart
sns.barplot(
    data=df,
    x='income_group',
    y='net_workforce_change_pct',
    order=income_order,
    errorbar=None,
    palette='coolwarm'
)

plt.title('Average Net Workforce Change % by Country Income Group', fontsize=14, fontweight='bold')
plt.xlabel('Income Group', fontsize=12)
plt.ylabel('Average Net Workforce Change %', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create a histogram with a KDE curve
sns.histplot(
    data=df,
    x='sector_automation_risk_score',
    bins=25,          # Number of intervals/bars
    kde=True,         # Adds the smooth trend line
    color='royalblue',
    alpha=0.7
)

# Customizing titles and labels
plt.title('Distribution of Sector Automation Risk Scores', fontsize=14, fontweight='bold')
plt.xlabel('Sector Automation Risk Score', fontsize=12)
plt.ylabel('Count', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Set the Seaborn theme and grab a Seaborn color palette
sns.set_theme(style="whitegrid")
colors = sns.color_palette('pastel') # Options: 'pastel', 'set2', 'muted', 'blues_r'

# 2. Prepare the data by counting the frequency of each group
income_counts = df['income_group'].value_counts()

# 3. Create the pie chart using Matplotlib
plt.figure(figsize=(8, 8))
plt.pie(
    income_counts,
    labels=income_counts.index,
    autopct='%1.1f%%',          # Displays percentages automatically
    startangle=140,             # Rotates the start point for a clean layout
    colors=colors,              # Applies the Seaborn color palette
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5} # Adds clean white borders
)

# Customizing title
plt.title('Proportion of Records by Country Income Group', fontsize=14, fontweight='bold')

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create a vertical violin plot
sns.violinplot(
    data=df,
    x='income_group',
    y='ai_adoption_index',
    palette='Set2',
    inner='quartile'  # Shows the median and quartiles inside the violin
)

# Customizing titles and labels
plt.title('Distribution of AI Adoption Index by Country Income Group', fontsize=14, fontweight='bold')
plt.xlabel('Income Group', fontsize=12)
plt.ylabel('AI Adoption Index', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#heatmap
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Select only the numerical columns from your dataframe
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# 2. Calculate the correlation matrix
corr_matrix = numeric_df.corr()

# 3. Plot the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(
    corr_matrix,
    annot=True,             # Shows the actual correlation numbers inside the squares
    cmap='coolwarm',        # Red for positive correlation, blue for negative
    fmt='.2f',              # Rounds the numbers to 2 decimal places
    linewidths=0.5,         # Adds a subtle white border between squares
    vmin=-1, vmax=1         # Sets the colorbar limits from -1 to 1
)

plt.title('Correlation Matrix of AI Workforce Metrics', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

# Create a step graph using lineplot with drawstyle
sns.lineplot(
    data=df,
    x='year',
    y='ai_adoption_index',
    drawstyle='steps-post',  # Vertical line happens AFTER the x-coordinate
    errorbar=None,
    marker='o',              # Adds points at the actual data steps
    linewidth=2.5,
    color='darkcyan'
)

plt.title('Step Graph: AI Adoption Index Trend (2020–2026)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average AI Adoption Index', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#boxplot

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

# Draw vertical boxplot over the years tracking workforce displacement %
sns.boxplot(
    data=df,
    x='year',
    y='pct_sector_workforce_displaced',
    color='skyblue',
    fliersize=4 # Controls the size of the outlier points
)

# Formatting
plt.title('Year-over-Year Global AI Workforce Displacement Distribution (2020–2026)', fontsize=14, pad=15)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Percentage Workforce Displaced (%)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.4, axis='y')

plt.tight_layout()
plt.show()
```

```python
#countplot
# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# 1. Get the sorted order of categories by frequency count
sector_counts = df['industry_sector'].value_counts().index

# 2. Create a horizontal countplot by passing the column to 'y'
sns.countplot(
    data=df,
    y='industry_sector',
    order=sector_counts,
    hue='industry_sector',  # Explicitly setting hue prevents deprecation warnings
    palette='viridis',
    legend=False
)

plt.title('Number of Records Available per Industry Sector', fontsize=14, fontweight='bold')
plt.xlabel('Count (Number of Rows)', fontsize=12)
plt.ylabel('Industry Sector', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#pairplot
# Set theme
sns.set_theme(style="whitegrid")

# 1. Select a clean subset of numeric metrics to compare
columns_to_plot = [
    'ai_adoption_index',
    'sector_automation_risk_score',
    'pct_sector_workforce_displaced',
    'industry_sector' # Kept for color-coding (hue)
]

# 2. Sample 500 rows randomly to prevent overplotting and memory crashes
df_sample = df[columns_to_plot].sample(n=500, random_state=42)

# 3. Create the pairplot
g = sns.pairplot(
    data=df_sample,
    hue='industry_sector',  # Colors the points by industry
    palette='tab10',
    corner=True,            # Optional: Removes the redundant top-right triangle of plots
    plot_kws={'alpha': 0.6, 's': 30} # Adjusts transparency and size of scatter points
)

# Customizing titles
g.figure.suptitle('Pairwise Relationships Across AI Workforce Metrics (Sampled)', fontsize=14, fontweight='bold', y=1.02)

plt.show()
```

```python
#comparison graph
# Set style
sns.set_theme(style="whitegrid")

# 1. Take a sample to prevent massive overplotting
df_sample = df.sample(n=1000, random_state=42)

# 2. Create the joint comparison graph
g = sns.jointplot(
    data=df_sample,
    x='ai_adoption_index',
    y='pct_sector_workforce_displaced',
    hue='industry_sector',  # Groups and colors both the scatter points and the marginal curves
    palette='tab10',
    alpha=0.7,
    height=8                # Jointplot manages its own figure size via height (square layout)
)

# Customizing titles (Jointplot handles titles via the underlying figure structure)
g.figure.suptitle('Comparing AI Adoption vs. Workforce Displacement Distributions', fontsize=14, fontweight='bold', y=1.02)

plt.show()
```

```python
#kde plot

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create a single variable KDE plot
sns.kdeplot(
    data=df,
    x='pct_sector_workforce_displaced',
    fill=True,
    color='purple',
    alpha=0.5,
    linewidth=2
)

plt.title('KDE Distribution of Global Workforce Displacement %', fontsize=14, fontweight='bold')
plt.xlabel('Percentage Workforce Displaced (%)', fontsize=12)
plt.ylabel('Density', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#strip plot
# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# 1. Take a sample of 1,000 rows to prevent extreme overplotting
df_sample = df.sample(n=1000, random_state=42)

# 2. Get the sorted order of income groups
income_order = df_sample['income_group'].value_counts().index

# 3. Create the strip plot
sns.stripplot(
    data=df_sample,
    x='income_group',
    y='pct_sector_workforce_displaced',
    order=income_order,
    hue='income_group',     # Sets color by category
    jitter=0.25,            # Spreads the dots out horizontally so density is visible
    size=5,                 # Size of the dots
    alpha=0.7,              # Adds transparency to see overlapping clusters
    palette='Set2',
    legend=False
)

plt.title('Distribution of Workforce Displacement % by Income Group (Sampled)', fontsize=14, fontweight='bold')
plt.xlabel('Income Group', fontsize=12)
plt.ylabel('Percentage Workforce Displaced (%)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
#jointplot
# Set style
sns.set_theme(style="whitegrid")

# 1. Take a sample to prevent massive overplotting
df_sample = df.sample(n=1000, random_state=42)

# 2. Create the joint plot
g = sns.jointplot(
    data=df_sample,
    x='ai_adoption_index',
    y='pct_sector_workforce_displaced',
    kind='scatter',          # Central plot is a scatter plot
    color='teal',
    alpha=0.6,
    height=8                # Jointplot manages its own figure size via 'height' (square layout)
)

# Customizing titles (Jointplot handles titles via the underlying figure structure)
g.figure.suptitle('Joint Plot: AI Adoption vs. Workforce Displacement (Sampled)', fontsize=14, fontweight='bold', y=1.02)

plt.show()
```

```python
#clustermap

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Isolate the numerical columns
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# 2. Calculate the correlation matrix
corr_matrix = numeric_df.corr()

# 3. Plot the Clustermap
g = sns.clustermap(
    corr_matrix,
    annot=True,             # Displays correlation coefficients inside the blocks
    cmap='vlag',            # Ideal diverging palette (blue for negative, red for positive)
    fmt='.2f',              # Rounds values to 2 decimal places
    vmin=-1, vmax=1,        # Set the bounds for the correlation scale
    linewidths=0.5,         # Adds a subtle border between squares
    figsize=(10, 10)
)

# Set the main title using the underlying figure object
g.figure.suptitle('Hierarchical Clustering of AI Workforce Metrics', fontsize=14, fontweight='bold', y=1.02)

plt.show()
```

```python
#relational plot
# Set style
sns.set_theme(style="whitegrid")

# 1. Take a sample to prevent overplotting across the grid
df_sample = df.sample(n=1000, random_state=42)

# 2. Create a faceted relational scatter plot
g = sns.relplot(
    data=df_sample,
    x='ai_adoption_index',
    y='pct_sector_workforce_displaced',
    hue='industry_sector',    # Colors points by industry sector
    col='income_group',       # Creates a separate horizontal subplot for each income group
    kind='scatter',           # Specifies a scatter relationship
    palette='tab10',
    alpha=0.7,
    height=5, aspect=1        # Controls the size/dimensions of each individual subplot square
)

# Customizing titles across the figure level
g.figure.suptitle('AI Adoption vs. Workforce Displacement Faceted by Income Group', fontsize=14, fontweight='bold', y=1.05)

plt.show()
```

```python
#categorical plot
# Set style
sns.set_theme(style="whitegrid")

# Create a faceted categorical boxplot grid
g = sns.catplot(
    data=df,
    x='year',
    y='pct_sector_workforce_displaced',
    col='income_group',       # Creates a separate column subplot for each income category
    kind='box',               # Specifies that the underlying plot should be a boxplot
    hue='year',               # Colors the boxes by year to prevent deprecation warnings
    palette='Blues',
    legend=False,
    height=5, aspect=0.8,     # Controls the height and width ratio of each subplot
    showfliers=False          # Hides extreme outliers to keep the grid comparison clean
)

# Customizing titles across the figure level
g.figure.suptitle('Workforce Displacement Distributions Faceted by Income Group', fontsize=14, fontweight='bold', y=1.05)
g.set_xticklabels(rotation=45) # Tilts year labels if they crowd together

plt.show()
```

```python
#distribution plot
# Set style
sns.set_theme(style="whitegrid")

# Create a figure-level continuous distribution plot
g = sns.displot(
    data=df,
    x='ai_adoption_index',
    hue='income_group',
    kind='kde',              # Specifies a smooth Kernel Density Estimate plot
    fill=True,               # Shades the areas under the curves
    palette='Set2',
    alpha=0.4,
    height=6, aspect=1.3     # Manages figure size directly inside displot
)

# Customizing titles and labels
g.figure.suptitle('Distribution of AI Adoption Index by Country Income Group', fontsize=14, fontweight='bold', y=1.02)
g.set_axis_labels('AI Adoption Index', 'Density', fontsize=12)

plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Generate mock data
np.random.seed(42)
students = ['Yadu','Elakkiya','Rithu', 'Muthu', 'Madhan','Sanu']
subjects = ["Math", "Science", "English", "History", "Art", "PE"]
marks = np.random.randint(40, 101, size=(6, 6))

# Build and reshape data for Seaborn
df = pd.DataFrame(marks, index=students, columns=subjects).reset_index()
df.rename(columns={'index': 'Student'}, inplace=True)
df_melted = pd.melt(df, id_vars=['Student'], var_name='Subject', value_name='Mark')

# Plotting
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

sns.barplot(x='Student', y='Mark', hue='Subject', data=df_melted, palette='muted')

plt.title('Student Performance Analysis across 6 Subjects', fontsize=14, pad=15)
plt.xlabel('Students', fontsize=12)
plt.ylabel('Marks Obtained', fontsize=12)
plt.legend(title='Subjects', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
