```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Group by the policy score (x) and get the average of y1, y2, and y3
# This clears out the overlapping points and gives us clean trend lines
grouped = df.groupby('govt_ai_policy_score_1_to_10')[
    ['ai_skill_wage_premium_pct', 'pct_sector_workforce_new_roles_created', 'gdp_per_capita_usd']
].mean().reset_index()

# Extract our x and y variables from the aggregated data
x_val = grouped['govt_ai_policy_score_1_to_10']
y1_val = grouped['ai_skill_wage_premium_pct']
y2_val = grouped['pct_sector_workforce_new_roles_created']
y3_val = grouped['gdp_per_capita_usd']

# 2. Normalize each variable to a 0-1 scale so they share the same graph beautifully
y1_norm = (y1_val - y1_val.min()) / (y1_val.max() - y1_val.min())
y2_norm = (y2_val - y2_val.min()) / (y2_val.max() - y2_val.min())
y3_norm = (y3_val - y3_val.min()) / (y3_val.max() - y3_val.min())

# 3. Create the single combined line graph
plt.figure(figsize=(10, 6))

# Plot each line with a unique color and label
plt.plot(x_val, y1_norm, label='AI Skill Wage Premium (y1)', color='royalblue', linewidth=2.5, marker='o')
plt.plot(x_val, y2_norm, label='New Roles Created % (y2)', color='forestgreen', linewidth=2.5, marker='s')
plt.plot(x_val, y3_norm, label='GDP per Capita (y3)', color='crimson', linewidth=2.5, marker='^')

# Customizing the chart elements
plt.xlabel('Government AI Policy Score (1 to 10)', fontsize=12)
plt.ylabel('Relative Growth / Trend Scale (Normalized 0 to 1)', fontsize=12)
plt.title('How AI Policy Scores Relate to Wages, Job Creation, and Wealth', fontsize=14, fontweight='bold')

# Add a legend to know which line is which
plt.legend(fontsize=11)

# Add a grid for better readability
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
