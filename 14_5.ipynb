```python


# Left Table (df1)
df1 = pd.DataFrame({'A': ['A1', 'A2', 'A3']}, index=[1, 2, 3])

# Right Table (df2)
df2 = pd.DataFrame({'B': ['B3', 'B4', 'B5']}, index=[3, 4, 5])

# How to apply them:
inner_res = df1.join(df2, how='inner') # Only Index 3
left_res  = df1.join(df2, how='left')  # Indices 1, 2, 3
right_res = df1.join(df2, how='right') # Indices 3, 4, 5
```

```python
inner_res
```

```python
left_res
```

```python
right_res
```
